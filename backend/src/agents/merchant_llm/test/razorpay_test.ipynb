```python
import os
import time
import razorpay
from dotenv import load_dotenv

load_dotenv()

KEY_ID = os.getenv("RAZORPAY_KEY_ID")
KEY_SECRET = os.getenv("RAZORPAY_KEY_SECRET")

client = razorpay.Client(auth=(KEY_ID, KEY_SECRET))

# Dynamically set close_by to 1 hour from current time
close_by_timestamp = int(time.time() + 3600)

# qr = client.qrcode.create({
#     "type": "upi_qr",
#     "name": "Store Front Display",
#     "usage": "single_use",
#     "fixed_amount": True,
#     "payment_amount": 30000,  # 30000 paise = ₹300
#     "description": "Test Payment",
#     "close_by": close_by_timestamp,
#     "notes": {
#         "purpose": "Test UPI QR Code"
#     }
# })

qr = client.qrcode.create({
  "type": "upi_qr",
  "name": "Store Front Display",
  "usage": "single_use",
  "fixed_amount": True,
  "payment_amount": 300,
  "description": "For Store 1",
  "customer_id": "cust_HKsR5se84c5LTO",
  "close_by": close_by_timestamp,
  "notes": {
    "purpose": "Test UPI QR Code notes"
  }
})

print(qr)
```
